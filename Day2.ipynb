```python
username = "sunny"
passwd = "sunny@11"
attempt = 0
while attempt < 3:
    un = input("Enter username: ")
    pw = input("Enter password: ")

    if un == username and pw == passwd:
        print("Login successful")
        break
    else:
        attempt += 1
        print("Incorrect credentials")
        print("Attempts left:", 3 - attempt)
```

```python
def palindrome(num1):
    n1 = num1
    rev = 0
    while num1>0:
        ldigit = num1%10
        rev = rev*10 + ldigit
        num1 = num1//10
    return "Palindrome" if rev==n1 else "Not Palindrome"
print(palindrome(1211))
```
