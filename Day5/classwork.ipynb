```python
class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age
        

class Teacher(Person):
    def __init__(self, name, age, address, phone, subjectname, experience):
        super().__init__(name, age, address, phone)
        self.subjectname = subjectname
        self.experience = experience


class Student(Person):
    def __init__(self, name, age, rollno):
        super().__init__(name, age)
        self.rollno = rollno


class Academic:
    def __init__(self, course, cgpa):
        self.course = course
        self.cgpa = cgpa


class Sports:
    def __init__(self, sport_name, level):
        self.sport_name = sport_name
        self.level = level


class AllRounderStudents(Student, Academic, Sports):
    def __init__(self, name, age, rollno, is_academic = False, is_sports = False, course = None, cgpa = None, sport_name = None, level = None):
        super().__init__(name, age, rollno)
        self.is_academic = is_academic
        self.is_sports = is_sports

        if self.is_academic:
            Academic.__init__(self, course, cgpa)

        if self.is_sports:
            Sports.__init__(self, sport_name, level)

    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Roll No: {self.rollno}")

        if self.is_academic:
            print(f"Course: {self.course}")
            print(f"CGPA: {self.cgpa}")

        if self.is_sports:
            print(f"Sport Name: {self.sport_name}")
            print(f"Sport Level: {self.level}")
```

```python
s1 = AllRounderStudents("Ankit", 22, 101, is_sports=True, sport_name="Cricket", level="Moderate")
s1.display()
s2 = AllRounderStudents("Manish", 22, 101, is_academic=True, course="ITI", cgpa=8.51)
s2.display()
```
