```python
from abc import ABC, abstractmethod

class BankAccount():
    def customer_name(self, name):
        self.name = name
        print("Customer Name: ", self.name)

    def date_of_birth(self, dob):
        self.dob = dob
        print("Date of Birth: ", self.dob)

    @abstractmethod
    def deposit(self, amount):
        pass
    @abstractmethod
    def withdraw(self, amount):
        pass

class SavingsAccount(BankAccount):
    min_balance = 1000
    balance = min_balance
    def deposit(self, amount):
        self.balance += amount
        print("Amount deposited: ", amount)
        print("Current balance: ", self.balance)
    
    def withdraw(self, amount):
        if self.balance - amount >= self.min_balance:
            self.balance -= amount
            print("Amount withdrawn: ", amount)
            print("Current balance: ", self.balance)
        else:
            print("Insufficient balance. Minimum balance must be maintained.")


class CurrentAccount(BankAccount):
    draft_limit = 7000
    balance = draft_limit

    def deposit(self, amount):
        self.balance += amount
        print("Amount deposited: ", amount)
        print("Current balance: ", self.balance)

    def withdraw(self, amount):
        if self.balance - amount >= self.draft_limit:
            self.balance -= amount
            print("Amount withdrawn: ", amount)
            print("Current balance: ", self.balance)

        if self.balance - amount == 0:
            self.balance -= amount
            print("Over Draft Amount Withdrawn: ", self.balance)
            print("Pay Over draft amount within next month or 5% charges will be applied after 20days compensation time!!!!!")



```

```python
customer1 = SavingsAccount()
customer1.deposit(5000)
customer1.withdraw(4000)
```
