#### **Matplotlib**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.random.randint(1, 50, 10)
y = np.random.randint(1, 50, 10)

print(x, y, sep = '\n')
```

```python
plt.plot(x, y)
```

```python
plt.plot(x, y, color='red', linestyle='dotted')
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.plot(x, y, color='red', linestyle='dotted')
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='dotted')
plt.show()
```

- -> solid(default)

-- -> dashed

-. -> dash-dot

: -> dotted

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='dashed')
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='-', alpha=0.3)
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='-', alpha=0.9,
         marker='*')
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='-', alpha=0.9,
         marker='*', markerfacecolor="y")
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='-', alpha=0.9,
         marker='*', markerfacecolor="y",
         markersize=10)
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='-', alpha=0.9,
         marker='*', markerfacecolor="y",
         markersize=10, markeredgecolor='black')
plt.show()
```

```python
plt.title('Random Chart')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis Label')
plt.xticks(x)
plt.yticks(y)
plt.plot(x, y, color='red', linestyle='-', alpha=0.9,
         marker='*', markerfacecolor="y",
         markersize=10, markeredgecolor='black',
         linewidth=5)
plt.show()
```

```python
x = np.arange(1, 11)
y = np.sin(x)
z = np.cos(x)

print(x, y, z, sep='\n\n')
```

```python
plt.plot(x, y, label = 'SINQ', color='red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.xticks()
plt.yticks()
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8, 4))
plt.plot(x, y, label = 'SINQ', color='red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.xticks(x)
plt.yticks(z)
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y, label = 'SINQ', color='red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.legend()


plt.subplot(1, 2, 2)
plt.plot(x, y, label = 'SINQ', color='green')
plt.plot(x, z, label = 'COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points2")
plt.ylabel("Associate Values2")
plt.legend()


plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.subplot(3, 3, 1)
plt.plot(x, y, label = 'SINQ', color='red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.legend()
plt.grid()


plt.subplot(3, 3, 5)
plt.plot(x, y, label = 'SINQ', color='green')
plt.plot(x, z, label = 'COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points2")
plt.ylabel("Associate Values2")
plt.legend()
plt.grid()

plt.subplot(3, 3, 9)
plt.plot(x, y, label = 'SINQ', color='green')
plt.plot(x, z, label = 'COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points3")
plt.ylabel("Associate Values3")
plt.subplots_adjust(wspace=0.7)
plt.legend()
plt.grid()


plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y, label = 'SINQ', color='red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.legend(loc='upper right')
plt.subplots_adjust(wspace=0.8)
plt.grid()


plt.subplot(1, 2, 2)
plt.plot(x, y, label = 'SINQ', color='green')
plt.plot(x, z, label = 'COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points2")
plt.ylabel("Associate Values2")
plt.legend(loc='upper right')
plt.subplots_adjust(wspace=0.8)
plt.grid()

plt.show()
```
