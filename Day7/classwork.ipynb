```python
with open('a.txt', 'w') as f:
    f.writelines('''gfgfliuhahuyuyvejbcbl \n
            tfegffba \n
            agfgfeajk \n
            bierievaj''')
```

```python
with open('a.txt', 'r+') as f:
    f.write('HIT')
```

```python
with open('data.txt') as f:
    words = f.read().split(',')
    print(len(words))
```

```python
with open('data.txt') as f:
    words = f.read().split()
    unique_words = set(words)
    print(unique_words)
    print(len(unique_words))
```

```python
with open('data.txt') as f:
    word_count = {}
    words = f.read().split()
    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1
    print(word_count)

```

```python
data = {
    "name": "Sunny",
    "age": 21,
    "course": "CS"
}
```

```python
import json
with open('json1.json', 'w') as f:
    json.dump(data, f)
```

```python
with open('json1.json', 'r') as f:
    # json.dump(data, f)
    print(json.load(f))
```

```python
import csv
with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Age", "Course"])
    writer.writerow(["Sunny", 21, "CS"])
```

```python
rows = [
    ["Name", "Age"],
    ["Piyush", 21],
    ["Rishu", 22]
]
with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(rows)
```

```python
with open("data.csv", "w", newline="") as file:
    fieldnames = ["name", "age"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"name": "Rishu", "age":22})
    writer.writerow({"name": "Piyush", "age": 21})
```

```python
with open("data.csv", "r") as file:
    reader = csv.DictReader(file)
    print(reader) # -> Object address
    for row in reader:
        print(row)
```
