```python
# Global Variable Access

x = 10

def show_value():
    print("Inside function:", x)

show_value()
print("Outside function:", x)
```

```python
# Modify Global Variable

count = 0

def increment():
    global count
    count += 1

increment()
print(count)
```

```python
# Without Global Keyword

count = 0

def increment():
    count += 1

increment()
print(count)
```

```python
# Local Variable

def demo():
    a = 5
    print(a)

demo()
```

```python
# Nonlocal Keyword

def outer():
    x = 10

    def inner():
        nonlocal x
        x += 5

    inner()
    print(x)

outer()
```
